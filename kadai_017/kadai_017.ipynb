```python
class Human:
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age

    def check_adult(self):
        if self.age >= 20:
            print(self.name, "は大人です。")
        else:
            print(self.name, "は大人ではありません。")


# Humanクラスのインスタンスをリストに追加
humans = []

humans.append(Human("田中太郎", 25))
humans.append(Human("佐藤花子", 18))
humans.append(Human("鈴木一郎", 20))

# リストの要素数分だけcheck_adultメソッドを呼び出す
for human in humans:
    human.check_adult()
```
